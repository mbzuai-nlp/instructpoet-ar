# Model Output Visualization

This notebook displays the model inputs and outputs side by side from the generation results.

```python
# Import required libraries
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load the TSV file
file_path = '/path/to/outputs/Qwen3-8B/curriculum/checkpoint-36000/generation/generation_ift_with_predictions.tsv'
df = pd.read_csv(file_path, sep='\t')

print(f"Loaded {len(df)} rows")
print(f"\nColumns: {list(df.columns)}")
print(f"\nDataFrame shape: {df.shape}")
```

```python
# Configure pandas display options for better viewing
pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)
```

```python
# Display input and output columns side by side
# Adjust column names based on what's in your file
# Common column names might be: 'input', 'output', 'prediction', 'generated_text', etc.

# Show available columns
print("Available columns:")
for col in df.columns:
    print(f"  - {col}")
```

```python
# Create a function to display examples side by side
def display_example(idx):
    """
    Display a single example with input, expected output, and model generation side by side
    """
    from IPython.display import display, HTML
    
    row = df.iloc[idx]
    
    # Try to identify input, output, and prediction columns
    input_cols = [col for col in df.columns if 'input' in col.lower() or 'prompt' in col.lower()]
    output_cols = [col for col in df.columns if 'output' in col.lower() and 'prediction' not in col.lower()]
    prediction_cols = [col for col in df.columns if 'prediction' in col.lower() or 'generated' in col.lower()]
    
    input_col = input_cols[0] if input_cols else df.columns[0]
    output_col = output_cols[0] if output_cols else (df.columns[1] if len(df.columns) > 1 else df.columns[0])
    prediction_col = prediction_cols[0] if prediction_cols else df.columns[-1]
    
    html = f"""
    <div style="display: flex; gap: 15px; margin-bottom: 20px;">
        <div style="flex: 1; border: 2px solid #4CAF50; padding: 15px; border-radius: 5px;">
            <h3 style="color: #4CAF50; margin-top: 0;">Input (Example {idx + 1})</h3>
            <div style="white-space: pre-wrap; font-family: monospace; direction: rtl; text-align: right;">{row[input_col]}</div>
        </div>
        <div style="flex: 1; border: 2px solid #FF9800; padding: 15px; border-radius: 5px;">
            <h3 style="color: #FF9800; margin-top: 0;">Expected Output (Example {idx + 1})</h3>
            <div style="white-space: pre-wrap; font-family: monospace; direction: rtl; text-align: right;">{row[output_col]}</div>
        </div>
        <div style="flex: 1; border: 2px solid #2196F3; padding: 15px; border-radius: 5px;">
            <h3 style="color: #2196F3; margin-top: 0;">Model Generation (Example {idx + 1})</h3>
            <div style="white-space: pre-wrap; font-family: monospace; direction: rtl; text-align: right;">{row[prediction_col]}</div>
        </div>
    </div>
    """
    
    display(HTML(html))
```

```python
# Display first 5 examples
import random
from IPython.display import display, HTML

display(HTML("<h2>First 5 Examples</h2>"))
for i in range(min(5, len(df))):
    idx = random.randint(0, len(df) - 1)
    display_example(idx)
```

```python
# Interactive widget to browse through examples
from ipywidgets import interact, IntSlider

@interact(example_idx=IntSlider(min=0, max=len(df)-1, step=1, value=0, description='Example:'))
def browse_examples(example_idx):
    display_example(example_idx)
```

```python
# Create a simple DataFrame view with selected columns
# This will show input, output, and prediction columns in a table format

# Identify columns
input_cols = [col for col in df.columns if 'input' in col.lower() or 'prompt' in col.lower()]
output_cols = [col for col in df.columns if 'output' in col.lower() and 'prediction' not in col.lower()]
prediction_cols = [col for col in df.columns if 'prediction' in col.lower() or 'generated' in col.lower()]

# Display columns to show
cols_to_show = []
if input_cols:
    cols_to_show.append(input_cols[0])
if output_cols:
    cols_to_show.append(output_cols[0])
if prediction_cols:
    cols_to_show.append(prediction_cols[0])

if cols_to_show:
    display_df = df[cols_to_show].head(10)
    display_df.style.set_properties(**{'text-align': 'right', 'direction': 'rtl'})
```
