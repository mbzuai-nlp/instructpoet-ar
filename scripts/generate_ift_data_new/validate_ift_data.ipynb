```python
import re
def remove_arabic_elongation(text: str) -> str:
    """
    Removes Arabic elongation characters (tatweel) and extra spaces.
    """
    # Remove tatweel
    text = text.replace("ـ", "")
    # Remove extra spaces
    text = re.sub(r" {2,}", " ", text).strip()
    return text

data[data['corrupted_poem'].isna()]
(data['corrupted_poem'] != data['poem_text']).sum()
```

```python
# Count samples where corrupted_poem differs from poem_text after removing elongation
# First, let's reload the corruption data to make sure we're working with the right dataset
data = pd.read_csv("/mnt/data/users/abdelrahman.sadallah/ONEDRIVE/poetry/final_test_data.csv")

# Apply elongation removal to both columns
data_clean = data.copy()
data_clean['poem_text_clean'] = data_clean['poem_text'].apply(lambda x: remove_arabic_elongation(str(x)) if pd.notna(x) else x)
data_clean['corrupted_poem_clean'] = data_clean['corrupted_poem'].apply(lambda x: remove_arabic_elongation(str(x)) if pd.notna(x) else x)

# Count differences after cleaning
different_after_cleaning = (data_clean['corrupted_poem_clean'] != data_clean['poem_text_clean']).sum()

print(f"Total samples: {len(data_clean)}")
print(f"Samples where corrupted_poem differs from poem_text (after removing elongation): {different_after_cleaning}")
print(f"Percentage of different samples: {(different_after_cleaning / len(data_clean)) * 100:.2f}%")

# Also check for any null values
null_corrupted = data_clean['corrupted_poem'].isna().sum()
null_original = data_clean['poem_text'].isna().sum()
print(f"\nNull values in corrupted_poem: {null_corrupted}")
print(f"Null values in poem_text: {null_original}")
```
