```python
import pandas as pd

# Read the TSV file
file_path = "/path/to/data/jais_data/de_dupped_train.tsv"
data = pd.read_csv(file_path, sep='\t')


print("total number of rows:", len(data))
# Filter rows where poem_verses is 1
filtered_data = data[data['poem_verses'] != 1]
print("total number of rows with poem_verses == 1:", len(filtered_data))
# Count the total number of words in the poem_text_without_diactarization column

```

```python
total_words = filtered_data['poem_text_no_diacritics'].str.split().str.len().sum()

print("Total number of words:", total_words)

average_words_per_poem = total_words / len(filtered_data)
print("Average number of words per poem:", average_words_per_poem)
average_verses_per_poem = filtered_data['poem_verses'].mean()
print("Average number of verses per poem:", average_verses_per_poem)
```

```python
from google import genai
import os
from dotenv import load_dotenv
# Load environment variables from the .env file
load_dotenv()
# Retrieve the Gemini API key
gemini_key = os.getenv("GEMINI_KEY_ANWAR")


client = genai.Client(api_key='gemini_key')
prompt = """
سـأجزيك بالقـد الـذي قد فككته	سأجزيك ما أبليتنا العام صعصعا
فــإن يـك محمـود أبـاك فإننـا	وجدناك منسوبا إلى الخير أروعا
سـأهدي وإن كنـا بتثليـث مدحـة	إليــك وإن حلـت بيوتـك لعلعـا
فـإن شـئت أهـدينا ثنـاء ومدحة	وإن شـئت عـدينا لكـم مئة معـا
"""

# Count the number of words in the prompt
prompt_word_count = len(prompt.split())
print("Number of words in the prompt:", prompt_word_count)
# Count tokens using the new client method.
total_tokens = client.models.count_tokens(
    model="gemini-2.5-flash", contents=prompt
)
# print("total_tokens: ", total_tokens)
# ( e.g., total_tokens: 10 )

response = client.models.generate_content(
    model="gemini-2.0-flash", contents=prompt
)

# The usage_metadata provides detailed token counts.
print(response.usage_metadata.candidates_token_count, 
      response.usage_metadata.prompt_token_count,
      response.usage_metadata.total_token_count)
# Print the response
print(response.candidates[0])
```
