```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
from concurrent.futures import ThreadPoolExecutor, as_completed
from tqdm import tqdm

# Get unique URLs from DataFrame
unique_urls = df["poet_page_url"].dropna().unique()

# Function to extract poet description from the precise container
def extract_poet_description(soup):
    container = soup.find("div", class_="col-md-7 col-lg-5 col-12 float-right")
    if container:
        h4_tag = container.find("h4", class_="main-color h6 mt-3")
        if h4_tag:
            return h4_tag.get_text(strip=True)
    return None

# Function to fetch and parse description from URL
def fetch_description(url):
    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()
        soup = BeautifulSoup(response.content, "html.parser")
        return url, extract_poet_description(soup)
    except Exception as e:
        return url, None

# Dictionary to store mapping
url_to_description = {}

# Use multithreading for speed
with ThreadPoolExecutor(max_workers=20) as executor:
    futures = {executor.submit(fetch_description, url): url for url in unique_urls}
    for future in tqdm(as_completed(futures), total=len(futures), desc="Fetching poet descriptions"):
        url, desc = future.result()
        url_to_description[url] = desc

```

```python
df2 = df.copy()

# Remove entries with empty values in url_to_description
url_to_description = {url: desc for url, desc in url_to_description.items() if desc}
# Update DataFrame only for rows with new poet descriptions
for url in df2["poet_page_url"].unique():
    if url in url_to_description:
        df2.loc[df2["poet_page_url"] == url, "poet_description"] = url_to_description[url]

df2[df2['poet_description'].isnull()]

```
