```python
import pandas as pd
import json

train_df = pd.read_csv('/path/to/data/cleaned_train.tsv', sep='\t')
test_df = pd.read_csv('/path/to/data/cleaned_test.tsv', sep='\t')
```

```python
train_df['dataset_name'].unique()
```

```python
import pandas as pd

# Filter out unwanted sources
excluded_sources = ['mawsooaa', 'boda_scrapped', 'ara_poet', 'adab_world']
filtered_sources = train_df[~train_df['dataset_name'].isin(excluded_sources)]

# Group by source and compute stats
source_stats = (
    filtered_sources
    .groupby('dataset_name', as_index=False)
    .agg(
        entries=('dataset_name', 'count'),
        avg_verses=('poem_verses', 'mean')
    )
    .sort_values(by='entries', ascending=False)
)

# Display source stats
print("Per-source statistics:")
print(source_stats.to_string(index=False, float_format="%.2f"))

# Compute overall stats
total_entries = source_stats['entries'].sum()
overall_avg_verses = (
    filtered_sources['poem_verses'].mean()
)

print("\nOverall statistics:")
print(f"Total entries: {total_entries}")
print(f"Overall average verses per poem: {overall_avg_verses:.2f}")
```
