```python
import pandas as pd
import json 

data_path = "<PROJECT_ROOT>/scrappers/adab/adab_with_descriptions.jsonl"

# Read the data from the JSONL file
with open(data_path, 'r') as file:
    data = [json.loads(line) for line in file]

# Convert the data to a DataFrame
df = pd.DataFrame(data)
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
from tqdm import tqdm

# Assume your DataFrame is called df
# Make sure poet_description is initialized
df['poet_description'] = df['poet_description'].fillna('')

# Dictionary to cache descriptions
poet_descriptions = {}

# Go through unique poet_page_urls
for url in tqdm(df['poet_page_url'].unique()):
    try:
        response = requests.get(url)
        soup = BeautifulSoup(response.content, 'html.parser')
        
        # Find the <p> tag with the class containing the biography
        p_tag = soup.find('p', class_='p-about p-about_biography')
        description = p_tag.get_text(separator="\n", strip=True) if p_tag else ''
        
        # Save to dictionary
        poet_descriptions[url] = description
    except Exception as e:
        print(f"Error processing {url}: {e}")
        poet_descriptions[url] = ''

# Map descriptions to the DataFrame
df['poet_description'] = df['poet_page_url'].map(poet_descriptions)
```
